```python
import matplotlib.pyplot as plt
import json

NB_THREAD_MAX = 16

data_file = open("./results.json", "r")

data = json.load(data_file)
```

```python
# Temps en fonction du nombre d'élément pour chacune des approche
times = []

data_im0 = data["image0.ppm"]
times.append([(int(nbT), data_im0[nbT]["time"]) for nbT in data_im0])

data_im1 = data["image1.ppm"]
times.append([(int(nbT), data_im1[nbT]["time"]) for nbT in data_im1])

data_im2 = data["image2.ppm"]
times.append([(int(nbT), data_im2[nbT]["time"]) for nbT in data_im2])
```

```python
# Calcules des accélérations

acc = []

time_0_base_speed = times[0][0][1]
acc.append([(int(nbT), time_0_base_speed / data_im0[nbT]["time"]) for nbT in data_im0])

time_1_base_speed = times[1][0][1]
acc.append([(int(nbT), time_1_base_speed / data_im1[nbT]["time"]) for nbT in data_im1])

time_2_base_speed = times[2][0][1]
acc.append([(int(nbT), time_2_base_speed / data_im2[nbT]["time"]) for nbT in data_im2])
```

```python
#plot times
im0_times = [t[1] for t in times[0]]

plt.bar(range(1,NB_THREAD_MAX+1) ,im0_times, label="_ignore")
plt.xlabel("Nombre de thread")
plt.ylabel("Temps en second")
# plt.title(f"Temps pris par l'implémentation \"réduction simple\" en fonction du nombre de thread pour {TEST_SIZE[-1]} éléments.")

plt.savefig("./images/im0_times.svg")
plt.show()
```

```python

#plot acc

im0_acc = [t[1] for t in acc[0]]

plt.bar(range(1,NB_THREAD_MAX+1) ,im0_acc, label="_ignore")
plt.xlabel("Nombre de thread")
plt.ylabel("Accélération")
# plt.title(f"Temps pris par l'implémentation \"réduction simple\" en fonction du nombre de thread pour {TEST_SIZE[-1]} éléments.")

plt.savefig("./images/im0_acc.svg")
plt.show()

print(max(im0_acc))
```
